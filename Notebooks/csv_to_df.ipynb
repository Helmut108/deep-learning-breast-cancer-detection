```python
from pathlib import Path
import shutil

source_root = Path(
    "/Volumes/Samsung_T7/Not_Sync/CBIS-DDSM-All-doiJNLP-zzWs5zfZ/cbis_ddsm"
)
samples_dir = Path("./data/samples")
# If exist_ok=True, FileExistsError isn't raised for an existent path.
# If parents=True, the path with a target directory and one or more non-existent parent directories can be created, not getting FileNotFoundError.
# Basically, True is set to it
# parents=True: Creates all missing parent folders automatically.
# exist_ok=True: Prevents Python from crashing if the folder already exists
samples_dir.mkdir(parents=True, exist_ok=True)

# Use the first 8 dataframe rows
for i, row in df.head(8).iterrows():
    # / is considered an concatenation operator for Path objects, which allows you to concatenate paths in a more readable way.
    series_path = source_root / row["folder"] / row["study"] / row["series"]

    # Each series folder contains one DICOM file
    dicom_file = next(series_path.glob("*.dcm"), None)

    if dicom_file is None:
        print(f"No DICOM file found: {series_path}")
        continue

    # Prefix with row index to prevent filename collisions
    destination = samples_dir / f"{i}_{dicom_file.name}"
    shutil.copy2(dicom_file, destination)

print("Done.")
```
